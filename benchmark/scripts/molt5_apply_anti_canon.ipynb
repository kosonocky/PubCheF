```python
import pandas as pd

# test_df = pd.read_csv("../../data/final_datasets/pubmed/preprocessed_propagated_1_hard/scaffold_split/test.csv")
test_df = pd.read_csv("../../data/opentargets/opentargets_smiles_indications_20240626.csv")
```

```python
import json
with open("rdkit_to_pubchem_canon.json", "r") as f:
    rdkit_to_pubchem = json.load(f)

conv_df = pd.DataFrame(rdkit_to_pubchem.items(), columns=["smiles", "pubchem_smiles"])

# left merge
test_df = test_df.merge(conv_df, left_on="smiles", right_on="smiles", how="left")
```

```python
# test_df.to_csv("chefv2_pubmed_scaffold_test_with_pubchem_smiles.csv", index=False)
test_df.to_csv("opentargets_smiles_indications_20240626_pubchem_smiles.csv", index=False)
```
